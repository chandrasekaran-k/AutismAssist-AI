# AutismAssist AI — Data Understanding & EDA

Explore the screening and support dataset, distributions, class balance, missing values, and relationships.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/asd_screening_dataset.csv")
df.shape
```

```python
df.head()
df.info()
df.describe(include="all").T
```

```python
print("Missing values:")
print(df.isna().sum().sort_values(ascending=False).head(20))
print("\nDuplicate rows:", df.duplicated().sum())
print("\nASD class balance:")
print(df["ASD_Status"].value_counts(normalize=True))
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="ASD_Status")
plt.title("ASD Status Distribution")
plt.show()
```

```python
behavior_cols = [
    "Eye_Contact","Response_to_Name","Communication_Level",
    "Joint_Attention","Sensory_Regulation","Motor_Skills",
    "ADL_Skills","Hyperactivity_Control","Social_Interaction"
]
df[behavior_cols].hist(figsize=(14,10))
plt.tight_layout()
plt.show()
```

## Initial observations

- The dataset contains child-level screening/support variables.
- `ASD_Status` is the screening target.
- Five therapy/support columns are multi-label outputs.
- The dataset should be treated as a portfolio/research dataset, not a clinically validated dataset.
